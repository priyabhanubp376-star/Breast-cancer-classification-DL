```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn.datasets
from sklearn.model_selection import train_test_split
```

```python
# loading the dataset
breast_cancer_dataset = sklearn.datasets.load_breast_cancer()
```

```python
print(breast_cancer_dataset)
```

```python
#loading the data to  a dataframe
data_frame = pd.DataFrame(breast_cancer_dataset.data, columns= breast_cancer_dataset.feature_names)
```

```python
# print the first 5 rows
data_frame.head()
```

```python
# adding the 'target' column to the dataframe
data_frame['label']=breast_cancer_dataset.target
```

```python
data_frame.tail()
```

```python
#no of rows and column in the dataset
data_frame.shape
```

```python
# getting some info about the data
data_frame.info()
```

```python
# checking missing value
data_frame.isnull().sum()
```

```python
#statistical measure about the data
data_frame.describe()
```

```python
#checking the distribution of target variable
data_frame['label'].value_counts()
```

1 ->Benign
2 ->Malignant

```python
data_frame.groupby('label').mean()
```

```python
x=data_frame.drop(columns='label',axis=1)
y=data_frame['label']
```

```python
print(x)
```

```python
print(y)
```

splitting the data into training data and testing data


```python
x_train, x_test, y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
print(x.shape,x_train.shape,x_test.shape)
```

```python
#Standardize the data
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train_std=scaler.fit_transform(x_train)
x_test_std=scaler.transform(x_test)
```

```python
print(x_train_std)
```

Building the neural network

```python
# importing tensorflow and keras
import tensorflow as tf
tf.random.set_seed(3)
from tensorflow import keras
```

```python
#setting up the layer of neural network
model=keras.Sequential([
                        keras.Input(shape=(30,)), # 30 input value lega
                        keras.layers.Flatten(), # conert 1d line
                        keras.layers.Dense(20,activation='relu'), # hidden layer
                        keras.layers.Dense(2,activation='sigmoid') #ouput jo 2 neuron dega
])
```

```python
#compile the neural network
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
#Training the neural network
history =model.fit(x_train_std,y_train,validation_split=0.1,epochs=10)
```

#visualize the data

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])

plt.title('model accuracy')
plt.ylabel('accuracy')
plt.xlabel('epoch')

plt.legend(['training data','validation data'],loc='lower right')
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])

plt.title('model loss')
plt.ylabel('loss')
plt.xlabel('epoch')

plt.legend(['training data','validation data'],loc='upper right')
```

Accuracy of the model on test data

```python
loss,accuracy=model.evaluate(x_test_std,y_test)
print(accuracy)
```

# argmax function-> numpy.argmax() returns the index position of the largest element in an array

```python
input_data=(20.57,17.77,132.9,1326,0.08474,0.07864,0.0869,0.07017,0.1812,0.05667,0.5435,0.7339,3.398,74.08,0.005225,0.01308,0.0186,0.0134,0.01389,0.003532,24.99,23.41,158.8,1956,0.1238,0.1866,0.2416,0.186,0.275,0.08902
)
#Data ko numpy array mein badalna
input_data_as_numpy_array = np.asarray(input_data)

# Ek sample ke liye reshape karna
input_data_reshaped = input_data_as_numpy_array.reshape(1, -1)

# Data ko standardize karna (scaler jo pehle fit kiya tha)
input_data_std = scaler.transform(input_data_reshaped)

# Prediction nikaalna
prediction = model.predict(input_data_std)
print(prediction)

prediction_label = [np.argmax(prediction)]
print(prediction_label)

if (prediction_label[0] == 0):
    print('The tumor is Malignant')
else:
    print('The tumor is Benign')
```
